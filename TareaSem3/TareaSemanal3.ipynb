## Bonus 1

```python
from pytc2.sistemas_lineales import analyze_sys, pretty_print_bicuad_omegayq, tf2sos_analog, pretty_print_lti
from pytc2.general import print_latex, a_equal_b_latex_s
import sympy as sp

##Bessel de 2do orden
s = sp.Symbol('s')

coth = sp.together(1/s + 1/(3/s))## saco la cotangente hiperbolica

P, Q = sp.fraction(coth)## separo denominador de numerador

T = sp.simplify(P.subs(s,0)/(P + Q))## armo la transferencia

num, den = sp.fraction(T)## separo numerador del denomidaor

num = list(map(float, sp.Poly(num,s).all_coeffs()))## saco coeficientes del num
den = list(map(float, sp.Poly(den,s).all_coeffs()))
print_latex(a_equal_b_latex_s('T2(s)', pretty_print_lti(num, den, displaystr=False)))

## Bessel 3er orden

coth = sp.together(1/s + 1/(3/s+1/(5/s)))## saco la cotangente hiperbolica

P, Q = sp.fraction(coth)## separo denominador de numerador

T = sp.simplify(P.subs(s,0)/(P + Q))## armo la transferencia

num, den = sp.fraction(T)## separo numerador del denomidaor

num = list(map(float, sp.Poly(num,s).all_coeffs()))## saco coeficientes del num
den = list(map(float, sp.Poly(den,s).all_coeffs()))
print_latex(a_equal_b_latex_s('T3(s)', pretty_print_lti(num, den, displaystr=False)))

##Bessel de 4to orden

coth = sp.together(1/s + 1/(3/s+1/(5/s+1/(7/s))))## saco la cotangente hiperbolica

P, Q = sp.fraction(coth)## separo denominador de numerador

T = sp.simplify(P.subs(s,0)/(P + Q))## armo la transferencia

num, den = sp.fraction(T)## separo numerador del denomidaor

num = list(map(float, sp.Poly(num,s).all_coeffs()))## saco coeficientes del num
den = list(map(float, sp.Poly(den,s).all_coeffs()))
print_latex(a_equal_b_latex_s('T4(s)', pretty_print_lti(num, den, displaystr=False)))
```
