## Parte B
### 1. Función principal de un VNA y aplicaciones
Un analizador vectorial de redes (VNA) es un instrumento de medición que permite caracterizar el comportamiento de un circuito o dispositivo en 
función de la frecuencia, midiendo tanto la amplitud como la fase de las señales reflejadas y transmitidas.

Para ello, utiliza los parámetros de dispersión o parámetros S, que permiten estudiar cómo se comporta una red ante señales de distintas frecuencias.

En particular, el NanoVNA V2 permite medir:
- S11: coeficiente de reflexión en el puerto de entrada. Permite determinar la adaptación de impedancias.
- S21: coeficiente de transmisión desde el puerto de entrada hacia el de salida. Permite determinar la ganancia o atenuación y el desplazamiento 
de fase introducidos por el dispositivo.
Su utilización resulta especialmente importante en aplicaciones de radiofrecuencia, como el diseño y ajuste de antenas, filtros, amplificadores y 
líneas de transmisión, donde es necesario conocer el comportamiento del circuito en un amplio rango de frecuencias.

### 2. Comparar respecto a un DSO GW Insteak
El NanoVNA presenta un rango de frecuencias considerablemente mayor, lo que lo hace especialmente útil para caracterizar circuitos de radiofrecuencia.

En cuanto al rango dinámico, el NanoVNA V2 Plus4 alcanza 90 dB en condiciones específicas de medición. Por otro lado, el GW Instek utiliza un sistema 
de adquisición de 8 bits, equivalente a una relación ideal de aproximadamente 50 dB para una señal senoidal de escala completa. Sin embargo, este valor
es una estimación teórica de cuantización y no representa directamente el rango dinámico real del osciloscopio, por lo que no corresponde compararlos 
como especificaciones equivalentes.

Finalmente, la diferencia de impedancias responde a las aplicaciones de cada instrumento. El NanoVNA utiliza una impedancia nominal de 50 Ω, habitual
en sistemas de radiofrecuencia, mientras que el osciloscopio presenta una impedancia de entrada elevada para reducir la carga sobre el circuito que se 
desea medir.

### 3. Investigar si el equipo cuenta con alguna medición en el dominio del tiempo.
Además de medir parámetros S en el dominio de la frecuencia, el NanoVNA V2 permite realizar mediciones en el dominio del tiempo mediante una función de transformación de los datos obtenidos en frecuencia.

Una de sus principales aplicaciones es la reflectometría en el dominio del tiempo (TDR), que permite identificar discontinuidades de impedancia en líneas de transmisión y determinar aproximadamente su ubicación.

Esta función resulta útil para detectar fallas en cables, como circuitos abiertos, cortocircuitos o cambios de impedancia.

El equipo permite obtener tres tipos de respuestas:
- Bandpass: simula la respuesta impulsiva del dispositivo.
- Low Pass Impulse: permite analizar la respuesta ante un impulso.
- Low Pass Step: simula una respuesta al escalón, útil para identificar discontinuidades de impedancia.
